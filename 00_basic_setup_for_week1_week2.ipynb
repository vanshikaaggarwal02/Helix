## 6) Verify Python Can Call Ollama (OpenAI-Compatible API)

Ollama exposes a local endpoint at `http://localhost:11434/`.

No real API key is needed for Ollama; we pass a dummy value because the client expects one.

````python
# Now you can use the OpenAI library to interact with the Ollama server. Make sure to run the PATH cell above to fix the PATH for the OpenAI library.       
from openai import OpenAI 

OLLAMA_BASE_URL = "http://localhost:11434/v1"

# Make a chat completion request to the Ollama server using the OpenAI library.
# Normally, this client talks to OpenAI’s servers.
# But here, we’re overriding the base URL to point to our local setup — where Ollama is running.
#  So instead of: Sending requests to OpenAI, We send requests to our local machine 
client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
response = client.chat.completions.create(
    model="llama3.2",
    messages=[{"role": "user", "content": "What is 2 + 2?"}],
)

print(response.choices[0].message.content)
# The model doesn’t always return just one answer — it returns a list of possible responses, hence we give [0] to get the first one.
````

# Gradio!

Today we will build User Interfaces using the outrageously simple Gradio framework.

Prepare for joy!

Please note: your Gradio screens may appear in 'dark mode' or 'light mode' depending on your computer settings.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
````

````python
import gradio as gr # oh yeah!
````

````python
# Let's wrap a call to GPT-4.1-mini in a simple function

system_message = "You are a helpful assistant"
OLLAMA_BASE_URL = "http://localhost:11434/v1"

client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
def message_ollama(prompt):
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = client.chat.completions.create(model="llama3.2", messages=messages)
    return response.choices[0].message.content
````

````python
# This can reveal the "training cut off", or the most recent date in the training data

message_ollama("What is today's date?")
````

## User Interface time!

## OK! Let's write a slightly better chat callback!

````python

OLLAMA_BASE_URL = "http://localhost:11434/v1"

client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = client.chat.completions.create(model="llama3.2", messages=messages)
    return response.choices[0].message.content
````

## First step: Have llama3.2 figure out which links are relevant

### Use a call to llama3.2 to read the links on a webpage, and respond in structured JSON.  
It should decide which links are relevant, and replace relative links such as "/about" with "https://company.com/about".  
We will use "one shot prompting" in which we provide an example of how it should respond in the prompt.

This is an excellent use case for an LLM, because it requires nuanced understanding. Imagine trying to code this without LLMs by parsing and analyzing the webpage - it would be very hard!

Sidenote: there is a more advanced technique called "Structured Outputs" in which we require the model to respond according to a spec. 

````python
link_system_prompt = """
You are provided with a list of links found on a webpage.
You are able to decide which of the links would be most relevant to learn linux. You have to only
include links related to linux learnings, no other links are to be included.
You should respond in JSON as in this example:

{
    "links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
    ]
}
"""
````

````python
from Scraper import fetch_website_links

def get_links_user_prompt(url):
    user_prompt = f"""
Here is the list of links on the website {url} -
Please decide which of these are relevant web links for learning linux, 
respond with the full https URL in JSON format.
Do not include Terms of Service, Privacy, email links.

Links (some might be relative links):

"""
    links = fetch_website_links(url)
    user_prompt += "\n".join(links)
    return user_prompt
````

````python
print(get_links_user_prompt("https://www.geeksforgeeks.org"))
````

````python

OLLAMA_BASE_URL = "http://localhost:11434/v1"

client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
def select_relevant_links(url):
    print(f"Selecting relevant links for {url} by calling ollama")
    response = client.chat.completions.create(
        model="llama3.2",
        messages=[
            {"role": "system", "content": link_system_prompt},
            {"role": "user", "content": get_links_user_prompt(url)}
        ],
        response_format={"type": "json_object"}
    )
    result = response.choices[0].message.content
    links = json.loads(result)
    print(f"Found {len(links['links'])} relevant links")
    return links
````

````python
select_relevant_links("https://www.geeksforgeeks.org")
````
